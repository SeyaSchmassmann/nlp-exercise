1. Count number of trainable parameters in $\text{BERT}_{\text{LARGE}}$

    From the [BERT paper](https://arxiv.org/pdf/1810.04805.pdf):
    - Vocabulary size: WordPiece $30'522$
    - Hidden size (Embedding Dimension): $1024$
    - Maximum sequence length: $512$
    - Number of layers: $24$
    - Number of attention heads: $16$

  
    **Embedding Layer:**
    | Layer | Shape | Parameters |
    |-------|-------|------------|
    | Token Embedding      | $30'522 \cdot 1024$ | $31'254'528$     |
    | Positional Embedding | $512 \cdot 1024$    | $524'288$        |
    | **Total**            |                     |  |


    **Transformer Block:**
    | Layer | Shape | Parameters |
    |-------|-------|------------|
    | Key            | $1024 \cdot 1024 + 1024$ | $1'048'576$ |
    | Query          | $1024 \cdot 1024 + 1024$ | $1'048'576$ |
    | Value          | $1024 \cdot 1024 + 1024$ | $1'048'576$ |
    | Output         | $1024 \cdot 1024 + 1024$ | $1'048'576$ |
    | Feed Forward 1 | $1024 \cdot 4096$        | $4'194'304$ |
    | Feed Forward 2 | $4096 \cdot 1024$        | $4'194'304$ |
    | Layer Norm 1   | $2 \cdot 1024$ \add      | $4'096$ |
    | Layer Norm 2   | $2 \cdot 1024$ \add      | $4'096$ |
    | *Total* | |  |
    | **Total** | $24 \cdot ...$ |  |
